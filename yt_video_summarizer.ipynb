```python
from youtube_transcript_api import YouTubeTranscriptApi
```

```python
def get_video_id(url_link):
    return url_link.split("watch?v=")[-1].split("&")[0]
```

```python
video_id = get_video_id("https://www.youtube.com/watch?v=xDQL3vWwcp0&t=161s")
```

```python
ytt_api = YouTubeTranscriptApi()
transcript = ytt_api.fetch(video_id)
```

```python
transcript
```

```python
transcript_joined = " ".join((line.text for line in transcript))
```

```python
transcript_joined
```

```python
from transformers import pipeline

# Load the same model directly on CPU (device=-1 forces CPU)
punct_pipe = pipeline(
    "token-classification",
    model="felflare/bert-restore-punctuation",
    aggregation_strategy="simple",
    device=-1
)
```

```python
result= punct_pipe(transcript_joined)
print(result)
```

```python
from google import genai
from google.colab import userdata
```

### Initialize the Gemini client using your API key directly

```python
client = genai.Client(api_key=userdata.get("GEMINI_API_KEY"))
```

```python
# Create your summarization prompt using your transcript variable
prompt = f'"Summerize this text: \ntext = "{transcript_joined}"'
```

```python
response = client.models.generate_content(
    model = "gemini-3.5-flash-lite",
    contents=prompt
)
print(response.text)
```
